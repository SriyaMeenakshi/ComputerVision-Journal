*Annotating Images*
We will learn how to perform the following annotations to the images

- Draw lines
- Draw circles
- Draw rectangles
- Add text



```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
#run configurations parameters for the plots
plt.rcParams['figure.figsize']=(9.0,9.0) #the default size of the plots is set to 9x9 inches
%matplotlib inline
```

```python
image=cv2.imread('images/space.jpg',cv2.IMREAD_COLOR)
plt.imshow(image[:,:,::-1])
```

Draw a line using cv2.line() function.

- Syntax img=cv2.line(img,pt1,pt2,color[,thickness[,lineType[,shift]]])
- pt1,pt2 are x,y coordinates of start and end points of line
- color of line to be drawn
- thickness is integer specifying thickness default is 1
- lineType type of line default is 8-connected line cv2.LINE_AA is used

```python
imgLine=image.copy()                #yellow
cv2.line(imgLine,(50,50),(250,50),(0,255,255),thickness=20,lineType=cv2.LINE_AA);
#line starts from (50,50) and ends at(250,50)
#the color yellow and opencv in bgr
#thickness is 5px
#lineType is cv2.LINE_AA
plt.imshow(imgLine[:,:,::-1])
```

Drawing a circle
- Syntax img=cv2.circle(img,center,radius,color[,thickness[,lineType[, shift]]])
- thickness positive outline,negative it is filled


```python
imgCircle=image.copy()                 #red
cv2.circle(imgCircle,(900,500),100,(0,0,255),thickness=5,lineType=cv2.LINE_AA);
plt.imshow(imgCircle[:,:,::-1])
```

Drawing a rectangle
- Syntax img=cv2.rectangle(img,pt1,pt2,color[,thickness[,lineType[,shift]]])
- pt1 is top-left vertex
- pt2 is opposite to pt1 on bottom right


```python
imgRectangle=image.copy()
cv2.rectangle(imgRectangle,(0,0),(200,500),(255,0,255),thickness=-5,lineType=cv2.LINE_AA);
plt.imshow(imgRectangle[:,:,::-1])
```

Adding text
- syntax img=cv2.putText(img,text,org,fontFace,fontScale,color[,thickness[, lineType[,bottomLeftOrigin]]])
- text to typed
- org is bottom left corner of the string in the image
- fontFace is font type
- fontScale that is multiplied by the font-specific base size
- color: Font color
- thickness is line thickness, default value is 1
- https://docs.opencv.org/4.5.1/d6/d6e/group__imgproc__draw.html#ga5126f47f883d730f633d74f07456c576


```python
imgText=image.copy()
text="Chandrayaan launch vehicle"
fontScale=2.3
fontFace=cv2.FONT_HERSHEY_SIMPLEX
fontColor=(0,0,0)
fontThickness=3
cv2.putText(imgText,text,(200,600),fontFace,fontScale,fontColor,fontThickness,cv2.LINE_AA)
plt.imshow(imgText[:,:,::-1])
```
